```python
word = input("Enter a word: ")
if word == word[::-1]:
    print("It's a palindrome!")
else:
    print("Not a palindrome.")
```

```python
word = input("Enter a word: ")

reversed_word = word[::-1]
print("Reversed string:", reversed_word)

if word == reversed_word:
    print("It's a palindrome!")
else:
    print("Not a palindrome.")
```
